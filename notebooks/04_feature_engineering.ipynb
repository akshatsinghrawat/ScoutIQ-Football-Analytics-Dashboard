Rather than relying only on raw statistics such as goals and assists, feature engineering helps create more meaningful performance indicators.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/players_data_cleaned.csv")

df.head()
```

## Creating New Features

The original dataset contains many useful statistics. However, we can derive additional metrics that better represent player performance.

These engineered features will be used throughout the remainder of the project.

## Feature 1 – Goal Difference

### Definition

Goal Difference measures the difference between the number of goals a player scored and their Expected Goals (xG).

### Formula

Goal Difference = Goals − Expected Goals (xG)

### Why it matters

Players with a positive Goal Difference have scored more goals than expected, indicating efficient finishing. Conversely, a negative value suggests that a player scored fewer goals than expected.

```python
df["Goal_Difference"] = df["Gls"] - df["xG"]

df[["Player", "Gls", "xG", "Goal_Difference"]].head()
```

### Result

A new feature named **Goal_Difference** has been successfully created. This metric will be used later to identify players who outperform or underperform their expected goal values.

## Feature 2 – Goal Contributions

### Definition

Goal Contributions represent the total number of goals and assists recorded by a player.

### Formula

Goal Contributions = Goals + Assists

### Why it matters

Some players contribute by scoring goals, while others contribute by creating chances for teammates. Goal Contributions combines both aspects into a single attacking metric, making it useful for comparing overall offensive impact.

```python
df["Goal_Contributions"] = df["Gls"] + df["Ast"]

df[["Player", "Gls", "Ast", "Goal_Contributions"]].head()
```

Some Other Players

```python
df.sort_values("Goal_Contributions", ascending=False)[
    ["Player", "Gls", "Ast", "Goal_Contributions"]
].head(10)
```

### Result

A new feature named **Goal_Contributions** has been created. This metric measures a player's overall attacking output by combining goals and assists into a single statistic.

## Feature 3 – Goals per 90 Minutes

### Definition

Goals per 90 Minutes estimates how many goals a player scores during a full 90-minute match.

### Formula

Goals per 90 = (Goals / Minutes Played) × 90

### Why it matters

Total goals alone can be misleading because players receive different amounts of playing time. Goals per 90 provides a fair comparison by standardizing performance over a full match.

```python
df["Goals_per90"] = ((df["Gls"] / df["Min"]) * 90).round(2)

df[["Player", "Gls", "Min", "Goals_per90"]].sample(10)
```

### Result

A new feature named **Goals_per90** has been created. This metric standardizes goal-scoring performance based on playing time, making it possible to compare players fairly regardless of the number of minutes they have played.

```python
df_filtered = df[df["Min"] >= 900]
```

## Feature 4 – Assists per 90 Minutes

### Definition

Assists per 90 Minutes estimates how many assists a player provides during a full 90-minute match.

### Formula

Assists per 90 = (Assists / Minutes Played) × 90

### Why it matters

Total assists do not account for differences in playing time. Assists per 90 provides a standardized measure of creativity, allowing fair comparisons between players who have played different numbers of minutes.

```python
df["Assists_per90"] = ((df["Ast"] / df["Min"]) * 90).round(2)

df[["Player", "Ast", "Min", "Assists_per90"]].sample(10)
```

### Result

A new feature named **Assists_per90** has been created. This metric measures a player's creative contribution on a per-match basis, making it easier to compare players with different amounts of playing time.

## Feature 5 – Goal Contributions per 90 Minutes

### Definition

Goal Contributions per 90 Minutes measures the average number of goals and assists a player contributes during a full 90-minute match.

### Formula

Goal Contributions per 90 = (Goal Contributions / Minutes Played) × 90

### Why it matters

This metric combines both goal scoring and chance creation into a single standardized measure. It provides a fair comparison between players regardless of playing time.

```python
df["GC_per90"] = (
    (df["Goal_Contributions"] / df["Min"]) * 90
).round(2)

df[
    [
        "Player",
        "Goal_Contributions",
        "Min",
        "GC_per90"
    ]
].sample(10)
```

### Result

A new feature named **GC_per90** has been created. This metric represents a player's total attacking contribution per full match and combines both goals and assists into a single standardized performance indicator.

## Feature 6 – Finishing Efficiency

### Definition

Finishing Efficiency measures how effectively a player converts expected goals into actual goals.

### Formula

Finishing Efficiency = Goals / Expected Goals (xG)

### Why it matters

A value greater than 1 indicates that a player scored more goals than expected, suggesting highly efficient finishing. A value below 1 indicates that the player scored fewer goals than expected.

```python
(df["xG"] == 0).sum()
```

```python
df["Finishing_Efficiency"] = np.where(
    df["xG"] > 0,
    (df["Gls"] / df["xG"]).round(2),
    np.nan
)

df[
    [
        "Player",
        "Gls",
        "xG",
        "Finishing_Efficiency"
    ]
].sample(10)
```

### Result

A new feature named **Finishing_Efficiency** has been created. To avoid division by zero, players with an Expected Goals (xG) value of zero were assigned a missing value (`NaN`). This ensures the dataset remains suitable for further analysis, SQL queries, Power BI dashboards, and machine learning models.

## Final Dataset Validation

Before exporting the dataset, we verify that all newly engineered features have been created successfully and contain valid values.

```python
new_features = [
    "Goal_Difference",
    "Goal_Contributions",
    "Goals_per90",
    "Assists_per90",
    "GC_per90",
    "Finishing_Efficiency"
]

df[new_features].describe()
```

```python
df[new_features].isnull().sum()
```
